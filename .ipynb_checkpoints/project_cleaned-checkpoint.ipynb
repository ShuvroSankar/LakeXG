```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error

# Load both CSV files
fcr = pd.read_csv("./Dataset/Chlorophyll_FCR.csv")
sugg = pd.read_csv("./Dataset/Chlorophyll_SUGG.csv")

# Combine into one dataframe
satellite = pd.concat([fcr, sugg], ignore_index=True)

# Convert date column (auto-detect format)
satellite['date'] = pd.to_datetime(satellite['date'])

# Remove negative chlorophyll values (keep >= 0)
satellite = satellite[satellite['chla_sat'] >= 0].copy()
print(f"Valid satellite dates after filtering: {len(satellite)}")

# # For each lake and date, average multiple tile values
# satellite = satellite.groupby(['lake', 'date'], as_index=False)['chla_sat'].mean()

# # Rename lake column to match in_situ
# satellite = satellite.rename(columns={'lake': 'lake_code'})

# print(satellite.head())
# print(f"Satellite data shape: {satellite.shape}")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from tqdm import tqdm
import gc

# ── 1. LOAD RAW PARQUET FILES ─────────────────────────────────────────────────
DATA_DIR = Path("./Dataset/LakeBeD-US-CSE")
REQUIRED_COLS = ['datetime', 'depth', 'temp', 'do', 'chla_ugl', 'par', 'fdom']

data_path = DATA_DIR / "Data"
lake_dirs = list(data_path.glob("HighFrequency/*"))

dataframes = []
for lake_dir in tqdm(lake_dirs, desc="Loading"):
    if not lake_dir.is_dir():
        continue
    lake_code = lake_dir.name
    for file_path in lake_dir.glob("*_2D.parquet"):
        try:
            df = pd.read_parquet(file_path)
            # Only keep columns that exist in this file
            cols_to_load = [c for c in REQUIRED_COLS if c in df.columns]
            df = df[cols_to_load].copy()
            df['lake_code'] = lake_code
            dataframes.append(df)
        except Exception as e:
            print(f"  Skipped {file_path.name}: {e}")
        gc.collect()

raw_df = pd.concat(dataframes, ignore_index=True, sort=False)
print(f"Raw rows loaded: {len(raw_df):,}")
print(f"Lakes: {raw_df['lake_code'].unique()}")

# ── 2. FILTER TO SURFACE DEPTH ≤ 0.5m ────────────────────────────────────────
raw_df = raw_df[raw_df['depth'] <= 0.5].copy()
print(f"After depth filter: {len(raw_df):,} rows")

# ── 3. AGGREGATE TO DAILY AVERAGES ────────────────────────────────────────────
raw_df['date'] = pd.to_datetime(raw_df['datetime']).dt.date

daily_shallow = (
    raw_df
    .groupby(['lake_code', 'date'])[['temp', 'do', 'chla_ugl', 'par', 'fdom']]
    .mean()
    .reset_index()
)

print(f"Daily rows: {len(daily_shallow):,}")
print("\nNon-null chla_ugl per lake:")
print(
    daily_shallow.groupby('lake_code')['chla_ugl']
    .count()
    .sort_values(ascending=False)
    .to_string()
)
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
import matplotlib.pyplot as plt
import shap

# ── CONFIGURATION ─────────────────────────────────────────────────────────────
LAKES_ALL = ['SUGG', 'BARC', 'LIRO', 'CRAM', 'PRPO', 'PRLA']
LAKES_WITH_SATELLITE = ['SUGG']  # add more as you collect GEE data

features = ['temp', 'do', 'par', 'fdom']
target = 'chla_ugl'

results = {}

for lake_code in LAKES_ALL:
    print(f"\n{'='*50}")
    print(f"Processing lake: {lake_code}")
    print('='*50)

    lake_df = daily_shallow[daily_shallow['lake_code'] == lake_code].copy()
    lake_df['date'] = pd.to_datetime(lake_df['date']).dt.date
    lake_df = lake_df.dropna(subset=[target])
    print(f"In-situ rows with chla_ugl: {len(lake_df)}")

    

    if len(lake_df) < 20:
        print(f"  Skipping {lake_code} — not enough data.")
        continue

    # Clip negatives then log-transform target
    lake_df[target] = np.log1p(lake_df[target].clip(lower=0))

    # Impute features
    imputer = SimpleImputer(strategy='median')
    lake_df = lake_df.copy()
    lake_df[features] = imputer.fit_transform(lake_df[features])

    # Lag + seasonal features with gap protection
    lake_df = lake_df.sort_values('date').reset_index(drop=True)
    lake_df['date_dt'] = pd.to_datetime(lake_df['date'])
    lake_df['day_gap'] = lake_df['date_dt'].diff().dt.days
    lake_df['chla_lag1'] = lake_df[target].shift(1)
    lake_df.loc[lake_df['day_gap'] > 3, 'chla_lag1'] = np.nan
    lake_df['chla_lag1'] = lake_df['chla_lag1'].fillna(lake_df[target].median())
    lake_df['month'] = lake_df['date_dt'].dt.month
    lake_df['dayofyear'] = lake_df['date_dt'].dt.dayofyear

    features_ext = features + ['chla_lag1', 'month', 'dayofyear']

    # Train/test split
    if lake_code in LAKES_WITH_SATELLITE:
        sat_lake = satellite[satellite['lake'] == lake_code].copy()
        sat_dates = set(sat_lake['date'].tolist())
        print(f"Valid satellite dates: {len(sat_dates)}")
        train_mask = ~lake_df['date'].isin(sat_dates)
        test_mask = lake_df['date'].isin(sat_dates)
    else:
        split_idx = int(len(lake_df) * 0.8)
        train_mask = lake_df.index < split_idx
        test_mask = lake_df.index >= split_idx

    X = lake_df[features_ext]
    y = lake_df[target]
    dates = lake_df['date']

    X_train, X_test = X[train_mask], X[test_mask]
    y_train, y_test = y[train_mask], y[test_mask]
    dates_test = dates[test_mask]

    print(f"Train size: {len(X_train)}, Test size: {len(X_test)}")

    if len(X_test) < 5:
        print(f"  Skipping {lake_code} — too few test rows.")
        continue

    # Train
    tscv = TimeSeriesSplit(n_splits=5)
    param_grid = {
        'n_estimators': [100, 200, 300],
        'max_depth': [3, 5, 7],
        'learning_rate': [0.01, 0.05, 0.1],
        'subsample': [0.7, 0.8, 0.9]
    }
    grid = GridSearchCV(
        xgb.XGBRegressor(random_state=42),
        param_grid, cv=tscv, scoring='r2', n_jobs=-1, verbose=0
    )
    grid.fit(X_train, y_train)
    best_model = grid.best_estimator_
    print(f"Best params: {grid.best_params_}")

    # Predict and back-transform
    y_pred_log = best_model.predict(X_test)
    y_pred = np.expm1(y_pred_log)       # back to µg/L
    y_test_actual = np.expm1(y_test)    # back to µg/L

    r2_insitu = r2_score(y_test_actual, y_pred)
    rmse_insitu = np.sqrt(mean_squared_error(y_test_actual, y_pred))
    print(f"XGBoost vs in-situ: R²={r2_insitu:.3f}, RMSE={rmse_insitu:.2f} µg/L")

    # Satellite comparison
    r2_sat = rmse_sat = bias_sat = merged = None

    if lake_code in LAKES_WITH_SATELLITE:
        pred_df = pd.DataFrame({'date': dates_test, 'chla_pred': y_pred})
        merged = pd.merge(pred_df, sat_lake[['date', 'chla_sat']], on='date', how='inner')
        print(f"Matched XGBoost-satellite days: {len(merged)}")

        if len(merged) >= 5:
            r2_sat = r2_score(merged['chla_sat'], merged['chla_pred'])
            rmse_sat = np.sqrt(mean_squared_error(merged['chla_sat'], merged['chla_pred']))
            bias_sat = (merged['chla_pred'] - merged['chla_sat']).mean()
            print(f"XGBoost vs Satellite: R²={r2_sat:.3f}, RMSE={rmse_sat:.2f} µg/L, Bias={bias_sat:+.2f} µg/L")

    results[lake_code] = {
        'r2_insitu': r2_insitu, 'rmse_insitu': rmse_insitu,
        'r2_sat': r2_sat, 'rmse_sat': rmse_sat, 'bias_sat': bias_sat,
        'merged': merged, 'model': best_model,
        'X_test': X_test, 'y_test': y_test_actual, 'y_pred': y_pred,
        'features': features_ext
    }

# Summary table
print("\n\n=== FINAL RESULTS SUMMARY ===")
print(f"{'Lake':<8} {'R²(in-situ)':>12} {'RMSE(µg/L)':>12} {'R²(sat)':>10} {'RMSE(sat)':>11} {'Bias(sat)':>10}")
print("-" * 68)
for lake, m in results.items():
    sat_r2   = f"{m['r2_sat']:.3f}"    if m['r2_sat']   is not None else "N/A"
    sat_rmse = f"{m['rmse_sat']:.2f}"  if m['rmse_sat'] is not None else "N/A"
    sat_bias = f"{m['bias_sat']:+.2f}" if m['bias_sat'] is not None else "N/A"
    print(f"{lake:<8} {m['r2_insitu']:>12.3f} {m['rmse_insitu']:>12.2f} {sat_r2:>10} {sat_rmse:>11} {sat_bias:>10}")

# ── SHAP + SCATTER PLOTS PER LAKE ─────────────────────────────────────────────
for lake_code, m in results.items():
    has_sat = m['merged'] is not None and len(m['merged']) >= 5

    fig, axes = plt.subplots(1, 2 if has_sat else 1,
                             figsize=(14 if has_sat else 6, 5))
    if not has_sat:
        axes = [axes]

    # SHAP plot
    plt.sca(axes[0])
    explainer = shap.TreeExplainer(m['model'])
    shap_values = explainer.shap_values(m['X_test'])
    shap.summary_plot(shap_values, m['X_test'], feature_names=m['features'],
                      show=False, plot_size=None)
    axes[0].set_title(f'SHAP Feature Importance — {lake_code}')

    # Satellite scatter (only if available)
    if has_sat:
        merged = m['merged']
        axes[1].scatter(merged['chla_sat'], merged['chla_pred'], alpha=0.7, edgecolors='k')
        max_val = max(merged['chla_sat'].max(), merged['chla_pred'].max())
        axes[1].plot([0, max_val], [0, max_val], 'r--', label='1:1 line')
        axes[1].set_xlabel('Satellite Chl-a (µg/L)')
        axes[1].set_ylabel('XGBoost Predicted Chl-a (µg/L)')
        axes[1].set_title(f'XGBoost vs Satellite — {lake_code}\n'
                          f'R²={m["r2_sat"]:.3f}, RMSE={m["rmse_sat"]:.2f}, Bias={m["bias_sat"]:+.2f}')
        axes[1].legend()
        axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(f'results_{lake_code}.png', dpi=150, bbox_inches='tight')
    plt.show()
    print(f"Saved: results_{lake_code}.png")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt

# Use daily_shallow (already filtered to depth <=0.5 and daily averaged)
sugg_daily = daily_shallow[daily_shallow['lake_code'] == 'SUGG'].copy()

# Keep only necessary columns (depth excluded)
features = ['temp', 'do', 'par', 'fdom']   # removed 'depth'
target = 'chla_ugl'

# Remove rows where target is missing
sugg_daily = sugg_daily.dropna(subset=[target])
print(f"SUGG daily rows with chla_ugl: {len(sugg_daily)}")

# Check feature coverage
print("Feature non‑null counts:")
print(sugg_daily[features].count())

# Impute missing feature values with median
imputer = SimpleImputer(strategy='median')
sugg_daily[features] = imputer.fit_transform(sugg_daily[features])

# Now we have a complete dataset
X = sugg_daily[features]
y = sugg_daily[target]
dates = sugg_daily['date']

# Identify the satellite dates for SUGG
sat_dates = satellite[satellite['lake_code'] == 'SUGG']['date'].unique()
print(f"Satellite dates for SUGG: {len(sat_dates)}")

# Split: train on all dates NOT in satellite dates, test on satellite dates
train_mask = ~dates.isin(sat_dates)
test_mask = dates.isin(sat_dates)

X_train, X_test = X[train_mask], X[test_mask]
y_train, y_test = y[train_mask], y[test_mask]
dates_train, dates_test = dates[train_mask], dates[test_mask]

print(f"Training size: {len(X_train)}, Test size (satellite days): {len(X_test)}")

# Train XGBoost
model = xgb.XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)

# Predict on test set
y_pred = model.predict(X_test)

# Evaluate predictions against in‑situ test values
r2_insitu = r2_score(y_test, y_pred)
rmse_insitu = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"XGBoost vs in‑situ (test days): R²={r2_insitu:.3f}, RMSE={rmse_insitu:.2f}")

# Merge predictions with satellite data
pred_df = pd.DataFrame({
    'lake_code': 'SUGG',
    'date': dates_test,
    'chla_pred': y_pred
})

sat_sugg = satellite[satellite['lake_code'] == 'SUGG'].copy()
merged_pred = pd.merge(pred_df, sat_sugg[['date', 'chla_sat']], on='date', how='inner')
print(f"Matched days for XGBoost vs satellite: {len(merged_pred)}")

if len(merged_pred) > 0:
    r2_sat = r2_score(merged_pred['chla_sat'], merged_pred['chla_pred'])
    rmse_sat = np.sqrt(mean_squared_error(merged_pred['chla_sat'], merged_pred['chla_pred']))
    bias_sat = (merged_pred['chla_pred'] - merged_pred['chla_sat']).mean()
    print(f"XGBoost vs Satellite: R²={r2_sat:.3f}, RMSE={rmse_sat:.2f} µg/L, Bias={bias_sat:.2f} µg/L")
    
    plt.figure(figsize=(6,6))
    plt.scatter(merged_pred['chla_sat'], merged_pred['chla_pred'], alpha=0.6, edgecolors='k')
    max_val = max(merged_pred['chla_sat'].max(), merged_pred['chla_pred'].max())
    plt.plot([0, max_val], [0, max_val], 'r--')
    plt.xlabel('Satellite Chla (µg/L)')
    plt.ylabel('XGBoost Predicted Chla (µg/L)')
    plt.title(f'XGBoost vs Satellite (SUGG): R²={r2_sat:.2f}, RMSE={rmse_sat:.2f}')
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    print("No matches – check date alignment.")
```

```python
import shap
import numpy as np
import pandas as pd

shap_rankings = {}

for lake_code, m in results.items():
    explainer = shap.TreeExplainer(m['model'])
    shap_values = explainer.shap_values(m['X_test'])
    
    mean_shap = np.abs(shap_values).mean(axis=0)
    
    # Rank features by mean absolute SHAP value
    ranking = pd.DataFrame({
        'feature': m['features'],
        'mean_shap': mean_shap
    }).sort_values('mean_shap', ascending=False).reset_index(drop=True)
    
    ranking['rank'] = ranking.index + 1
    shap_rankings[lake_code] = ranking
    
    print(f"\n=== {lake_code} ===")
    print(ranking[['rank', 'feature', 'mean_shap']].to_string(index=False))

# ── Cross-lake summary table ──────────────────────────────────────────────────
features_all = results[list(results.keys())[0]]['features']
summary = pd.DataFrame({'feature': features_all})

for lake_code, ranking in shap_rankings.items():
    rank_map = dict(zip(ranking['feature'], ranking['rank']))
    summary[lake_code] = summary['feature'].map(rank_map)

summary = summary.set_index('feature')
print("\n\n=== CROSS-LAKE SHAP RANKING TABLE ===")
print(summary.to_string())
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# BASELINE MODEL COMPARISON
# Drop this cell immediately after the main XGBoost training cell.
# Requires: daily_shallow, satellite, results  (all produced by the main pipeline)
# Adds to results[lake]: 'r2_persist', 'rmse_persist', 'r2_ridge', 'rmse_ridge'
# ═══════════════════════════════════════════════════════════════════════════════

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error

# ── CONFIG (must match main pipeline exactly) ──────────────────────────────────
LAKES_ALL            = ['SUGG', 'BARC', 'LIRO', 'CRAM', 'PRPO', 'PRLA']
LAKES_WITH_SATELLITE = ['SUGG', 'BARC', 'LIRO', 'CRAM']
features     = ['temp', 'do', 'par', 'fdom']
target       = 'chla_ugl'
features_ext = features + ['chla_lag1', 'month', 'dayofyear']

# ── BASELINE LOOP ──────────────────────────────────────────────────────────────
for lake_code in LAKES_ALL:
    print(f"\n{'='*50}")
    print(f"Baselines: {lake_code}")
    print('='*50)

    # ── 1. Rebuild the exact same preprocessed DataFrame ──────────────────────
    lake_df = daily_shallow[daily_shallow['lake_code'] == lake_code].copy()
    lake_df['date'] = pd.to_datetime(lake_df['date']).dt.date
    lake_df = lake_df.dropna(subset=[target])

    if len(lake_df) < 20:
        print(f"  Skipping {lake_code} — not enough data.")
        continue

    # Log-transform target (identical to main pipeline)
    lake_df[target] = np.log1p(lake_df[target].clip(lower=0))

    # Median imputation on sensor features
    imputer = SimpleImputer(strategy='median')
    lake_df[features] = imputer.fit_transform(lake_df[features])

    # Lag + seasonal features with gap protection
    lake_df = lake_df.sort_values('date').reset_index(drop=True)
    lake_df['date_dt']  = pd.to_datetime(lake_df['date'])
    lake_df['day_gap']  = lake_df['date_dt'].diff().dt.days
    lake_df['chla_lag1'] = lake_df[target].shift(1)
    lake_df.loc[lake_df['day_gap'] > 3, 'chla_lag1'] = np.nan
    lake_df['chla_lag1'] = lake_df['chla_lag1'].fillna(lake_df[target].median())
    lake_df['month']     = lake_df['date_dt'].dt.month
    lake_df['dayofyear'] = lake_df['date_dt'].dt.dayofyear

    # ── 2. Same train/test split logic ────────────────────────────────────────
    if lake_code in LAKES_WITH_SATELLITE:
        sat_lake  = satellite[satellite['lake_code'] == lake_code].copy()
        sat_dates = set(sat_lake['date'].dt.date.tolist())  # ensure date objects match
        train_mask = ~lake_df['date'].isin(sat_dates)
        test_mask  =  lake_df['date'].isin(sat_dates)
    
        # Fallback to 80/20 if satellite dates give too few test rows
        if test_mask.sum() < 5:
            print(f"  Satellite dates too sparse — falling back to 80/20 split.")
            split_idx  = int(len(lake_df) * 0.8)
            train_mask = lake_df.index < split_idx
            test_mask  = lake_df.index >= split_idx
    else:
        split_idx  = int(len(lake_df) * 0.8)
        train_mask = lake_df.index < split_idx
        test_mask  = lake_df.index >= split_idx

    X      = lake_df[features_ext]
    y      = lake_df[target]

    X_train, X_test = X[train_mask], X[test_mask]
    y_train, y_test = y[train_mask], y[test_mask]

    if len(X_test) < 5:
        print(f"  Skipping {lake_code} — too few test rows.")
        continue

    # Back-transform helper (identical to main pipeline)
    y_test_actual = np.expm1(y_test)

    # ── 3. BASELINE A: Persistence (predict today = yesterday) ────────────────
    # chla_lag1 IS the persistence prediction (already in log space, back-transform)
    y_persist = np.expm1(X_test['chla_lag1'].values)

    r2_persist   = r2_score(y_test_actual, y_persist)
    rmse_persist = np.sqrt(mean_squared_error(y_test_actual, y_persist))
    print(f"Persistence  — R²={r2_persist:.3f}, RMSE={rmse_persist:.2f} µg/L")

    # ── 4. BASELINE B: Ridge regression (same 7 features) ────────────────────
    ridge_pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('ridge',  Ridge(alpha=1.0))
    ])
    ridge_pipe.fit(X_train, y_train)
    y_ridge_log = ridge_pipe.predict(X_test)
    y_ridge     = np.expm1(y_ridge_log)

    r2_ridge   = r2_score(y_test_actual, y_ridge)
    rmse_ridge = np.sqrt(mean_squared_error(y_test_actual, y_ridge))
    print(f"Ridge        — R²={r2_ridge:.3f}, RMSE={rmse_ridge:.2f} µg/L")

    # ── 5. XGBoost result from main pipeline ──────────────────────────────────
    r2_xgb   = results[lake_code]['r2_insitu']
    rmse_xgb = results[lake_code]['rmse_insitu']
    print(f"XGBoost      — R²={r2_xgb:.3f}, RMSE={rmse_xgb:.2f} µg/L")

    # Store back into results dict for the summary table
    results[lake_code]['r2_persist']   = r2_persist
    results[lake_code]['rmse_persist'] = rmse_persist
    results[lake_code]['r2_ridge']     = r2_ridge
    results[lake_code]['rmse_ridge']   = rmse_ridge


# ── SUMMARY TABLE ─────────────────────────────────────────────────────────────
print("\n\n=== BASELINE COMPARISON SUMMARY ===")
print(f"{'Lake':<6} {'Persist R²':>11} {'Persist RMSE':>13} "
      f"{'Ridge R²':>9} {'Ridge RMSE':>11} "
      f"{'XGB R²':>7} {'XGB RMSE':>9}")
print("-" * 72)
for lake in LAKES_ALL:
    if lake not in results or 'r2_persist' not in results[lake]:
        continue
    m = results[lake]
    print(f"{lake:<6} "
          f"{m['r2_persist']:>11.3f} {m['rmse_persist']:>13.2f} "
          f"{m['r2_ridge']:>9.3f} {m['rmse_ridge']:>11.2f} "
          f"{m['r2_insitu']:>7.3f} {m['rmse_insitu']:>9.2f}")

# ── GROUPED BAR CHART ─────────────────────────────────────────────────────────
lakes_plot = [l for l in LAKES_ALL if l in results and 'r2_persist' in results[l]]

r2_persist_vals = [results[l]['r2_persist']   for l in lakes_plot]
r2_ridge_vals   = [results[l]['r2_ridge']     for l in lakes_plot]
r2_xgb_vals     = [results[l]['r2_insitu']    for l in lakes_plot]

x     = np.arange(len(lakes_plot))
width = 0.25

fig, ax = plt.subplots(figsize=(9, 4))

bars1 = ax.bar(x - width, r2_persist_vals, width, label='Persistence',
               color='#B4B2A9', edgecolor='white', linewidth=0.5)
bars2 = ax.bar(x,          r2_ridge_vals,   width, label='Ridge regression',
               color='#5DCAA5', edgecolor='white', linewidth=0.5)
bars3 = ax.bar(x + width,  r2_xgb_vals,     width, label='XGBoost (LakeXG)',
               color='#1D9E75', edgecolor='white', linewidth=0.5)

# Value labels on bars
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        h = bar.get_height()
        va = 'bottom' if h >= 0 else 'top'
        offset = 0.01 if h >= 0 else -0.01
        ax.text(bar.get_x() + bar.get_width() / 2,
                h + offset, f'{h:.2f}',
                ha='center', va=va, fontsize=7.5, color='#2C2C2A')

ax.set_xticks(x)
ax.set_xticklabels(lakes_plot, fontsize=10)
ax.set_ylabel('R²', fontsize=10)
ax.set_title('Model comparison: Persistence vs Ridge vs XGBoost (LakeXG)',
             fontsize=11, fontweight='normal', pad=10)
ax.axhline(0, color='#888780', linewidth=0.8, linestyle='--')
ax.legend(fontsize=9, framealpha=0.9)
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.2f'))
ax.set_ylim(min(min(r2_persist_vals), 0) - 0.1, 1.05)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', alpha=0.25, linewidth=0.5)

plt.tight_layout()
plt.savefig('baseline_comparison.png', dpi=180, bbox_inches='tight')
plt.show()
print("Saved: baseline_comparison.png")
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# BASELINE MODEL COMPARISON
# Drop this cell immediately after the main XGBoost training cell.
# Requires: daily_shallow, satellite, results  (all produced by the main pipeline)
# Adds to results[lake]: 'r2_persist', 'rmse_persist', 'r2_ridge', 'rmse_ridge'
# ═══════════════════════════════════════════════════════════════════════════════

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error

# ── CONFIG (must match main pipeline exactly) ──────────────────────────────────
LAKES_ALL            = ['SUGG', 'BARC', 'LIRO', 'CRAM', 'PRPO', 'PRLA']
LAKES_WITH_SATELLITE = ['SUGG', 'BARC', 'LIRO', 'CRAM']
features     = ['temp', 'do', 'par', 'fdom']
target       = 'chla_ugl'
features_ext = features + ['chla_lag1', 'month', 'dayofyear']

print("=== Training data size per lake ===")
for lake_code in LAKES_ALL:
    lake_df = daily_shallow[daily_shallow['lake_code'] == lake_code].copy()
    lake_df = lake_df.dropna(subset=['chla_ugl'])
    split_idx = int(len(lake_df) * 0.8)
    print(f"{lake_code}: total={len(lake_df)}, train={split_idx}, test={len(lake_df)-split_idx}")

# ── BASELINE LOOP ──────────────────────────────────────────────────────────────
for lake_code in LAKES_ALL:
    print(f"\n{'='*50}")
    print(f"Baselines: {lake_code}")
    print('='*50)

    # ── 1. Rebuild the exact same preprocessed DataFrame ──────────────────────
    lake_df = daily_shallow[daily_shallow['lake_code'] == lake_code].copy()
    lake_df['date'] = pd.to_datetime(lake_df['date']).dt.date
    lake_df = lake_df.dropna(subset=[target])
    

    if len(lake_df) < 500:
            param_grid = {
                'n_estimators': [100, 200],
                'max_depth': [3, 4],
                'learning_rate': [0.05, 0.1],
                'subsample': [0.8]
                }
    else:
        param_grid = {
            'n_estimators': [100, 200, 300],
            'max_depth': [3, 5, 7],
            'learning_rate': [0.01, 0.05, 0.1],
            'subsample': [0.7, 0.8, 0.9]
            }

    # Log-transform target (identical to main pipeline)
    lake_df[target] = np.log1p(lake_df[target].clip(lower=0))

    # Median imputation on sensor features
    imputer = SimpleImputer(strategy='median')
    lake_df[features] = imputer.fit_transform(lake_df[features])

    # Lag + seasonal features with gap protection
    lake_df = lake_df.sort_values('date').reset_index(drop=True)
    lake_df['date_dt']  = pd.to_datetime(lake_df['date'])
    lake_df['day_gap']  = lake_df['date_dt'].diff().dt.days
    lake_df['chla_lag1'] = lake_df[target].shift(1)
    lake_df.loc[lake_df['day_gap'] > 3, 'chla_lag1'] = np.nan
    lake_df['chla_lag1'] = lake_df['chla_lag1'].fillna(lake_df[target].median())
    lake_df['month']     = lake_df['date_dt'].dt.month
    lake_df['dayofyear'] = lake_df['date_dt'].dt.dayofyear

    # ── 2. Same train/test split logic ────────────────────────────────────────
    if lake_code in LAKES_WITH_SATELLITE:
        sat_lake  = satellite[satellite['lake_code'] == lake_code].copy()
        sat_dates = set(sat_lake['date'].dt.date.tolist())  # ensure date objects match
        train_mask = ~lake_df['date'].isin(sat_dates)
        test_mask  =  lake_df['date'].isin(sat_dates)
    
        # Fallback to 80/20 if satellite dates give too few test rows
        if test_mask.sum() < 5:
            print(f"  Satellite dates too sparse — falling back to 80/20 split.")
            split_idx  = int(len(lake_df) * 0.8)
            train_mask = lake_df.index < split_idx
            test_mask  = lake_df.index >= split_idx
    else:
        split_idx  = int(len(lake_df) * 0.8)
        train_mask = lake_df.index < split_idx
        test_mask  = lake_df.index >= split_idx

    X      = lake_df[features_ext]
    y      = lake_df[target]

    X_train, X_test = X[train_mask], X[test_mask]
    y_train, y_test = y[train_mask], y[test_mask]

    if len(X_test) < 5:
        print(f"  Skipping {lake_code} — too few test rows.")
        continue

    # Back-transform helper (identical to main pipeline)
    y_test_actual = np.expm1(y_test)

    # ── 3. BASELINE A: Persistence (predict today = yesterday) ────────────────
    # chla_lag1 IS the persistence prediction (already in log space, back-transform)
    y_persist = np.expm1(X_test['chla_lag1'].values)

    r2_persist   = r2_score(y_test_actual, y_persist)
    rmse_persist = np.sqrt(mean_squared_error(y_test_actual, y_persist))
    print(f"Persistence  — R²={r2_persist:.3f}, RMSE={rmse_persist:.2f} µg/L")

    # ── 4. BASELINE B: Ridge regression (same 7 features) ────────────────────
    ridge_pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('ridge',  Ridge(alpha=1.0))
    ])
    ridge_pipe.fit(X_train, y_train)
    y_ridge_log = ridge_pipe.predict(X_test)
    y_ridge     = np.expm1(y_ridge_log)
    
    r2_ridge   = r2_score(y_test_actual, y_ridge)
    rmse_ridge = np.sqrt(mean_squared_error(y_test_actual, y_ridge))
    print(f"Ridge        — R²={r2_ridge:.3f}, RMSE={rmse_ridge:.2f} µg/L")
    # ── 5. XGBoost result from main pipeline ──────────────────────────────────
    r2_xgb   = results[lake_code]['r2_insitu']
    rmse_xgb = results[lake_code]['rmse_insitu']
    print(f"XGBoost      — R²={r2_xgb:.3f}, RMSE={rmse_xgb:.2f} µg/L")

    # Store back into results dict for the summary table
    results[lake_code]['r2_persist']   = r2_persist
    results[lake_code]['rmse_persist'] = rmse_persist
    results[lake_code]['r2_ridge']     = r2_ridge
    results[lake_code]['rmse_ridge']   = rmse_ridge


# ── SUMMARY TABLE ─────────────────────────────────────────────────────────────
print("\n\n=== BASELINE COMPARISON SUMMARY ===")
print(f"{'Lake':<6} {'Persist R²':>11} {'Persist RMSE':>13} "
      f"{'Ridge R²':>9} {'Ridge RMSE':>11} "
      f"{'XGB R²':>7} {'XGB RMSE':>9}")
print("-" * 72)
for lake in LAKES_ALL:
    if lake not in results or 'r2_persist' not in results[lake]:
        continue
    m = results[lake]
    print(f"{lake:<6} "
          f"{m['r2_persist']:>11.3f} {m['rmse_persist']:>13.2f} "
          f"{m['r2_ridge']:>9.3f} {m['rmse_ridge']:>11.2f} "
          f"{m['r2_insitu']:>7.3f} {m['rmse_insitu']:>9.2f}")

# ── GROUPED BAR CHART ─────────────────────────────────────────────────────────
lakes_plot = [l for l in LAKES_ALL if l in results and 'r2_persist' in results[l]]

r2_persist_vals = [results[l]['r2_persist']   for l in lakes_plot]
r2_ridge_vals   = [results[l]['r2_ridge']     for l in lakes_plot]
r2_xgb_vals     = [results[l]['r2_insitu']    for l in lakes_plot]

x     = np.arange(len(lakes_plot))
width = 0.25

fig, ax = plt.subplots(figsize=(9, 4))

bars1 = ax.bar(x - width, r2_persist_vals, width, label='Persistence',
               color='#B4B2A9', edgecolor='white', linewidth=0.5)
bars2 = ax.bar(x,          r2_ridge_vals,   width, label='Ridge regression',
               color='#5DCAA5', edgecolor='white', linewidth=0.5)
bars3 = ax.bar(x + width,  r2_xgb_vals,     width, label='XGBoost (LakeXG)',
               color='#1D9E75', edgecolor='white', linewidth=0.5)

# Value labels on bars
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        h = bar.get_height()
        va = 'bottom' if h >= 0 else 'top'
        offset = 0.01 if h >= 0 else -0.01
        ax.text(bar.get_x() + bar.get_width() / 2,
                h + offset, f'{h:.2f}',
                ha='center', va=va, fontsize=7.5, color='#2C2C2A')

ax.set_xticks(x)
ax.set_xticklabels(lakes_plot, fontsize=10)
ax.set_ylabel('R²', fontsize=10)
ax.set_title('Model comparison: Persistence vs Ridge vs XGBoost (LakeXG)',
             fontsize=11, fontweight='normal', pad=10)
ax.axhline(0, color='#888780', linewidth=0.8, linestyle='--')
ax.legend(fontsize=9, framealpha=0.9)
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.2f'))
ax.set_ylim(min(min(r2_persist_vals), 0) - 0.1, 1.05)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', alpha=0.25, linewidth=0.5)

plt.tight_layout()
plt.savefig('baseline_comparison.png', dpi=180, bbox_inches='tight')
plt.show()
print("Saved: baseline_comparison.png")
```
