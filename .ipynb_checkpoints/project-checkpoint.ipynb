```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, r2_score
import joblib
```

```python
# Load and impute Kaggle data
df_kaggle = pd.read_csv('./Dataset/water_potability.csv')
df_kaggle.head()
```

```python
imputer = SimpleImputer(strategy='median')
X_kaggle = pd.DataFrame(imputer.fit_transform(df_kaggle.drop('Potability', axis=1)),
                        columns=df_kaggle.drop('Potability', axis=1).columns)
y_kaggle = df_kaggle['Potability']
```

```python
# Train source classifier
X_train_k, X_test_k, y_train_k, y_test_k = train_test_split(X_kaggle, y_kaggle, test_size=0.2, random_state=42)
source_clf = RandomForestClassifier(n_estimators=100, random_state=42)
source_clf.fit(X_train_k, y_train_k)   # <-- removed extra ')'

print(f"Source model accuracy: {source_clf.score(X_test_k, y_test_k):.3f}")
print("Source feature importances:")
print(pd.Series(source_clf.feature_importances_, index=X_kaggle.columns).sort_values(ascending=False))

# Save model and imputer
joblib.dump(source_clf, 'source_model.pkl')   # <-- changed source_model to source_clf
joblib.dump(imputer, 'kaggle_imputer.pkl')
```

```python
# AQUAIR already loaded as df_aquair
# Features and target
features_aquair = ['temp', 'humid', 'co2', 'voc', 'pm25', 'pm10']
X_aquair = df_aquair[features_aquair].copy()
y_aquair = df_aquair['score'].copy()

# Quick check for missing values
print(f"Missing in X_aquair: {X_aquair.isnull().sum().sum()}")
print(f"Missing in y_aquair: {y_aquair.isnull().sum()}")

# If any missing, fill with median (should be none)
X_aquair = X_aquair.fillna(X_aquair.median())
y_aquair = y_aquair.fillna(y_aquair.median())

# Train/test split
X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(X_aquair, y_aquair, test_size=0.2, random_state=42)
```

```python
baseline_rf = RandomForestRegressor(n_estimators=100, random_state=42)
baseline_rf.fit(X_train_a, y_train_a)

y_pred_base = baseline_rf.predict(X_test_a)
mae_base = mean_absolute_error(y_test_a, y_pred_base)
r2_base = r2_score(y_test_a, y_pred_base)

print("=== Baseline (No Transfer) ===")
print(f"MAE: {mae_base:.2f}")
print(f"R2 Score: {r2_base:.3f}")
```

```python
# Get average tree depth from source classifier
source_depths = [tree.tree_.max_depth for tree in source_clf.estimators_]
avg_depth = int(np.mean(source_depths))

# Get min_samples_leaf from the source classifier (it's a parameter)
source_min_samples_leaf = source_clf.min_samples_leaf  # default is 1

# Transferred regressor with same complexity parameters
transferred_rf = RandomForestRegressor(
    n_estimators=100,
    max_depth=avg_depth,
    min_samples_leaf=source_min_samples_leaf,
    random_state=42
)
transferred_rf.fit(X_train_a, y_train_a)

y_pred_trans = transferred_rf.predict(X_test_a)
mae_trans = mean_absolute_error(y_test_a, y_pred_trans)
r2_trans = r2_score(y_test_a, y_pred_trans)

print("=== Transferred Model (using source tree depth & leaf size) ===")
print(f"MAE: {mae_trans:.2f}")
print(f"R2 Score: {r2_trans:.3f}")
```

```python
print("\n=== Improvement ===")
print(f"MAE reduction: {mae_base - mae_trans:.2f} ({(mae_base - mae_trans)/mae_base*100:.1f}%)")
if r2_trans > r2_base:
    print(f"R2 improved by {r2_trans - r2_base:.3f}")
else:
    print(f"R2 decreased by {r2_base - r2_trans:.3f}")

# Feature importance comparison between baseline and transferred
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].barh(features_aquair, baseline_rf.feature_importances_)
axes[0].set_title("Baseline Feature Importances")
axes[1].barh(features_aquair, transferred_rf.feature_importances_)
axes[1].set_title("Transferred Feature Importances")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from tqdm import tqdm
import gc

# ================= CONFIGURATION =================
DATA_DIR = Path("./Dataset/LakeBeD-US-CSE")
LOAD_2D_ONLY = True   # Keep as True for now

# Actual columns that exist in the 2D files (from your inspection)
REQUIRED_COLS = ['datetime', 'depth', 'temp', 'do', 'chla_ugl', 'par', 'fdom']

# ================= LOAD LAKE INFO =================
lake_info_path = DATA_DIR / "Data" / "Lake_Info.csv"
lake_info = pd.read_csv(lake_info_path)
print("Loaded lake info for:", lake_info['lake_name'].tolist())

# ================= FIND ALL PARQUET FILES =================
data_path = DATA_DIR / "Data"
lake_dirs = list(data_path.glob("HighFrequency/*")) + list(data_path.glob("LowFrequency/*"))

parquet_files = []
for lake_dir in lake_dirs:
    if not lake_dir.is_dir():
        continue
    lake_code = lake_dir.name
    for file_path in lake_dir.glob("*.parquet"):
        if LOAD_2D_ONLY and "_2D.parquet" not in file_path.name:
            continue
        parquet_files.append((file_path, lake_code))

print(f"Found {len(parquet_files)} Parquet files to load")

# ================= LOAD FILES WITH COLUMN SELECTION =================
dataframes = []
for file_path, lake_code in tqdm(parquet_files, desc="Loading"):
    try:
        df = pd.read_parquet(file_path, columns=REQUIRED_COLS)
        df['lake_code'] = lake_code
        dataframes.append(df)
    except Exception as e:
        print(f"Warning: Could not load {file_path.name} - {e}")
        continue
    
    # Periodic garbage collection
    if len(dataframes) % 10 == 0:
        gc.collect()

# ================= COMBINE ALL DATA =================
if not dataframes:
    raise RuntimeError("No data files were loaded. Check file paths and column names.")

combined_df = pd.concat(dataframes, ignore_index=True, sort=False)
print(f"\n✅ Combined data shape: {combined_df.shape}")
print(f"Columns: {combined_df.columns.tolist()}")
print(f"Memory usage: {combined_df.memory_usage(deep=True).sum() / 1e6:.2f} MB")

# ================= FILTER TO SURFACE MEASUREMENTS =================
# For each lake, find the minimum depth (usually 0 or the shallowest sensor)
combined_df['depth_min_per_lake'] = combined_df.groupby('lake_code')['depth'].transform('min')
surface_df = combined_df[combined_df['depth'] == combined_df['depth_min_per_lake']].copy()
surface_df.drop(columns=['depth_min_per_lake'], inplace=True)

print(f"\n🌊 Surface measurements: {surface_df.shape[0]} rows (from {combined_df.shape[0]} total)")
print(f"Unique depths kept: {surface_df['depth'].unique()}")

# ================= AGGREGATE TO DAILY AVERAGES =================
# Convert datetime to date (timezone-naive for grouping)
surface_df['date'] = surface_df['datetime'].dt.date

# Group by lake and date, taking mean of numerical columns
daily_avg = surface_df.groupby(['lake_code', 'date'])[['temp', 'do', 'chla_ugl', 'par', 'fdom']].mean().reset_index()

print(f"\n📅 Daily aggregated data shape: {daily_avg.shape}")
print(daily_avg.head())

# Optional: Save to CSV for later use
# daily_avg.to_csv("lakebed_daily_surface.csv", index=False)
```

```python
# After you have combined_df, define a function to get surface rows per lake
def get_surface_rows(df):
    # Keep only depths >= 0 (ignore negative or unrealistic depths)
    df_nonneg = df[df['depth'] >= 0].copy()
    if df_nonneg.empty:
        return pd.DataFrame()  # no valid surface data
    
    # For each lake, find the smallest depth that has at least one non-NaN value for key parameters
    # We'll use 'chla_ugl' as the primary target (or 'temp' if chla missing)
    key_col = 'chla_ugl' if 'chla_ugl' in df.columns else 'temp'
    
    # Group by lake and depth, count non-NaN in key_col
    depth_stats = df_nonneg.groupby(['lake_code', 'depth'])[key_col].count().reset_index()
    # For each lake, keep the shallowest depth where count > 0
    shallowest = depth_stats[depth_stats[key_col] > 0].groupby('lake_code')['depth'].min().reset_index()
    shallowest.rename(columns={'depth': 'surface_depth'}, inplace=True)
    
    # Merge back to get all rows at that depth
    surface_df = df_nonneg.merge(shallowest, on='lake_code')
    surface_df = surface_df[surface_df['depth'] == surface_df['surface_depth']].drop(columns=['surface_depth'])
    return surface_df

surface_df = get_surface_rows(combined_df)
print(f"Surface rows: {len(surface_df)}")
print(surface_df[['lake_code', 'depth']].drop_duplicates())
```

```python
import earthaccess
import geopandas as gpd
from shapely.geometry import Point
import pandas as pd

# 1. Authenticate with NASA
earthaccess.login(strategy="interactive", persist=True)

# 2. Define your lakes with precise coordinates
lakes = {
    'FCR': {'lat': 37.3047263, 'lon': -79.8383679},
    'SUGG': {'lat': 29.68705, 'lon': -82.01617}
}

# 3. Get the date range from your in-situ data
# Make sure in_situ is defined (uncomment the next line if needed)
# in_situ = daily_clean.copy()
dates = in_situ['date']
start_date = dates.min().strftime('%Y-%m-%d')
end_date = dates.max().strftime('%Y-%m-%d')

# 4. Search and download data
for lake_name, coords in lakes.items():
    # Create a point and buffer
    point = gpd.GeoSeries([Point(coords['lon'], coords['lat'])], crs='EPSG:4326')
    # Reproject to UTM zone 17N for accurate 500m buffer
    buffer = point.to_crs('EPSG:32617').buffer(500).to_crs('EPSG:4326')
    # Extract bounding box as (minx, miny, maxx, maxy)
    bbox = buffer.total_bounds  # returns (minx, miny, maxx, maxy)
    
    # Search for both Sentinel-2A and Sentinel-2B
    for sat in ['S2A_L2A_TILE', 'S2B_L2A_TILE']:
        results = earthaccess.search_data(
            short_name=sat,
            bounding_box=(bbox[0], bbox[1], bbox[2], bbox[3]),  # correct format
            temporal=(start_date, end_date),
            cloud_cover=(0, 20)
        )
        print(f"Found {len(results)} scenes for {lake_name} ({sat})")
        
        # Download first 5 (or fewer)
        if results:
            earthaccess.download(results[:5], f"./satellite_data/{lake_name}")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from tqdm import tqdm
import gc

# ── 1. LOAD RAW PARQUET FILES ─────────────────────────────────────────────────
DATA_DIR = Path("./Dataset/LakeBeD-US-CSE")
REQUIRED_COLS = ['datetime', 'depth', 'temp', 'do', 'chla_ugl', 'par', 'fdom']

data_path = DATA_DIR / "Data"
lake_dirs = list(data_path.glob("HighFrequency/*"))

dataframes = []
for lake_dir in tqdm(lake_dirs, desc="Loading"):
    if not lake_dir.is_dir():
        continue
    lake_code = lake_dir.name
    for file_path in lake_dir.glob("*_2D.parquet"):
        try:
            df = pd.read_parquet(file_path)
            # Only keep columns that exist in this file
            cols_to_load = [c for c in REQUIRED_COLS if c in df.columns]
            df = df[cols_to_load].copy()
            df['lake_code'] = lake_code
            dataframes.append(df)
        except Exception as e:
            print(f"  Skipped {file_path.name}: {e}")
        gc.collect()

raw_df = pd.concat(dataframes, ignore_index=True, sort=False)
print(f"Raw rows loaded: {len(raw_df):,}")
print(f"Lakes: {raw_df['lake_code'].unique()}")

# ── 2. FILTER TO SURFACE DEPTH ≤ 0.5m ────────────────────────────────────────
raw_df = raw_df[raw_df['depth'] <= 0.5].copy()
print(f"After depth filter: {len(raw_df):,} rows")

# ── 3. AGGREGATE TO DAILY AVERAGES ────────────────────────────────────────────
raw_df['date'] = pd.to_datetime(raw_df['datetime']).dt.date

daily_shallow = (
    raw_df
    .groupby(['lake_code', 'date'])[['temp', 'do', 'chla_ugl', 'par', 'fdom']]
    .mean()
    .reset_index()
)

print(f"Daily rows: {len(daily_shallow):,}")
print("\nNon-null chla_ugl per lake:")
print(
    daily_shallow.groupby('lake_code')['chla_ugl']
    .count()
    .sort_values(ascending=False)
    .to_string()
)
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
import matplotlib.pyplot as plt
import shap

# ── CONFIGURATION ─────────────────────────────────────────────────────────────
LAKES_ALL = ['SUGG', 'BARC', 'LIRO', 'CRAM', 'PRPO', 'PRLA']
LAKES_WITH_SATELLITE = ['SUGG']  # add more as you collect GEE data

features = ['temp', 'do', 'par', 'fdom']
target = 'chla_ugl'

results = {}

for lake_code in LAKES_ALL:
    print(f"\n{'='*50}")
    print(f"Processing lake: {lake_code}")
    print('='*50)

    lake_df = daily_shallow[daily_shallow['lake_code'] == lake_code].copy()
    lake_df['date'] = pd.to_datetime(lake_df['date']).dt.date
    lake_df = lake_df.dropna(subset=[target])
    print(f"In-situ rows with chla_ugl: {len(lake_df)}")

    if len(lake_df) < 20:
        print(f"  Skipping {lake_code} — not enough data.")
        continue

    # Clip negatives then log-transform target
    lake_df[target] = np.log1p(lake_df[target].clip(lower=0))

    # Impute features
    imputer = SimpleImputer(strategy='median')
    lake_df = lake_df.copy()
    lake_df[features] = imputer.fit_transform(lake_df[features])

    # Lag + seasonal features with gap protection
    lake_df = lake_df.sort_values('date').reset_index(drop=True)
    lake_df['date_dt'] = pd.to_datetime(lake_df['date'])
    lake_df['day_gap'] = lake_df['date_dt'].diff().dt.days
    lake_df['chla_lag1'] = lake_df[target].shift(1)
    lake_df.loc[lake_df['day_gap'] > 3, 'chla_lag1'] = np.nan
    lake_df['chla_lag1'] = lake_df['chla_lag1'].fillna(lake_df[target].median())
    lake_df['month'] = lake_df['date_dt'].dt.month
    lake_df['dayofyear'] = lake_df['date_dt'].dt.dayofyear

    features_ext = features + ['chla_lag1', 'month', 'dayofyear']

    # Train/test split
    if lake_code in LAKES_WITH_SATELLITE:
        sat_lake = satellite[satellite['lake'] == lake_code].copy()
        sat_dates = set(sat_lake['date'].tolist())
        print(f"Valid satellite dates: {len(sat_dates)}")
        train_mask = ~lake_df['date'].isin(sat_dates)
        test_mask = lake_df['date'].isin(sat_dates)
    else:
        split_idx = int(len(lake_df) * 0.8)
        train_mask = lake_df.index < split_idx
        test_mask = lake_df.index >= split_idx

    X = lake_df[features_ext]
    y = lake_df[target]
    dates = lake_df['date']

    X_train, X_test = X[train_mask], X[test_mask]
    y_train, y_test = y[train_mask], y[test_mask]
    dates_test = dates[test_mask]

    print(f"Train size: {len(X_train)}, Test size: {len(X_test)}")

    if len(X_test) < 5:
        print(f"  Skipping {lake_code} — too few test rows.")
        continue

    # Train
    tscv = TimeSeriesSplit(n_splits=5)
    param_grid = {
        'n_estimators': [100, 200, 300],
        'max_depth': [3, 5, 7],
        'learning_rate': [0.01, 0.05, 0.1],
        'subsample': [0.7, 0.8, 0.9]
    }
    grid = GridSearchCV(
        xgb.XGBRegressor(random_state=42),
        param_grid, cv=tscv, scoring='r2', n_jobs=-1, verbose=0
    )
    grid.fit(X_train, y_train)
    best_model = grid.best_estimator_
    print(f"Best params: {grid.best_params_}")

    # Predict and back-transform
    y_pred_log = best_model.predict(X_test)
    y_pred = np.expm1(y_pred_log)       # back to µg/L
    y_test_actual = np.expm1(y_test)    # back to µg/L

    r2_insitu = r2_score(y_test_actual, y_pred)
    rmse_insitu = np.sqrt(mean_squared_error(y_test_actual, y_pred))
    print(f"XGBoost vs in-situ: R²={r2_insitu:.3f}, RMSE={rmse_insitu:.2f} µg/L")

    # Satellite comparison
    r2_sat = rmse_sat = bias_sat = merged = None

    if lake_code in LAKES_WITH_SATELLITE:
        pred_df = pd.DataFrame({'date': dates_test, 'chla_pred': y_pred})
        merged = pd.merge(pred_df, sat_lake[['date', 'chla_sat']], on='date', how='inner')
        print(f"Matched XGBoost-satellite days: {len(merged)}")

        if len(merged) >= 5:
            r2_sat = r2_score(merged['chla_sat'], merged['chla_pred'])
            rmse_sat = np.sqrt(mean_squared_error(merged['chla_sat'], merged['chla_pred']))
            bias_sat = (merged['chla_pred'] - merged['chla_sat']).mean()
            print(f"XGBoost vs Satellite: R²={r2_sat:.3f}, RMSE={rmse_sat:.2f} µg/L, Bias={bias_sat:+.2f} µg/L")

    results[lake_code] = {
        'r2_insitu': r2_insitu, 'rmse_insitu': rmse_insitu,
        'r2_sat': r2_sat, 'rmse_sat': rmse_sat, 'bias_sat': bias_sat,
        'merged': merged, 'model': best_model,
        'X_test': X_test, 'y_test': y_test_actual, 'y_pred': y_pred,
        'features': features_ext
    }

# Summary table
print("\n\n=== FINAL RESULTS SUMMARY ===")
print(f"{'Lake':<8} {'R²(in-situ)':>12} {'RMSE(µg/L)':>12} {'R²(sat)':>10} {'RMSE(sat)':>11} {'Bias(sat)':>10}")
print("-" * 68)
for lake, m in results.items():
    sat_r2   = f"{m['r2_sat']:.3f}"    if m['r2_sat']   is not None else "N/A"
    sat_rmse = f"{m['rmse_sat']:.2f}"  if m['rmse_sat'] is not None else "N/A"
    sat_bias = f"{m['bias_sat']:+.2f}" if m['bias_sat'] is not None else "N/A"
    print(f"{lake:<8} {m['r2_insitu']:>12.3f} {m['rmse_insitu']:>12.2f} {sat_r2:>10} {sat_rmse:>11} {sat_bias:>10}")

# ── SHAP + SCATTER PLOTS PER LAKE ─────────────────────────────────────────────
for lake_code, m in results.items():
    has_sat = m['merged'] is not None and len(m['merged']) >= 5

    fig, axes = plt.subplots(1, 2 if has_sat else 1,
                             figsize=(14 if has_sat else 6, 5))
    if not has_sat:
        axes = [axes]

    # SHAP plot
    plt.sca(axes[0])
    explainer = shap.TreeExplainer(m['model'])
    shap_values = explainer.shap_values(m['X_test'])
    shap.summary_plot(shap_values, m['X_test'], feature_names=m['features'],
                      show=False, plot_size=None)
    axes[0].set_title(f'SHAP Feature Importance — {lake_code}')

    # Satellite scatter (only if available)
    if has_sat:
        merged = m['merged']
        axes[1].scatter(merged['chla_sat'], merged['chla_pred'], alpha=0.7, edgecolors='k')
        max_val = max(merged['chla_sat'].max(), merged['chla_pred'].max())
        axes[1].plot([0, max_val], [0, max_val], 'r--', label='1:1 line')
        axes[1].set_xlabel('Satellite Chl-a (µg/L)')
        axes[1].set_ylabel('XGBoost Predicted Chl-a (µg/L)')
        axes[1].set_title(f'XGBoost vs Satellite — {lake_code}\n'
                          f'R²={m["r2_sat"]:.3f}, RMSE={m["rmse_sat"]:.2f}, Bias={m["bias_sat"]:+.2f}')
        axes[1].legend()
        axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(f'results_{lake_code}.png', dpi=150, bbox_inches='tight')
    plt.show()
    print(f"Saved: results_{lake_code}.png")
```

```python
# Count non-null chla_ugl per lake
chla_counts = daily_shallow.groupby('lake_code')['chla_ugl'].count()
valid_lakes = chla_counts[chla_counts >= 30].index.tolist()
print("Lakes with ≥30 daily chla_ugl (depth ≤0.5):", valid_lakes)

# Filter
in_situ_shallow = daily_shallow[daily_shallow['lake_code'].isin(valid_lakes)].dropna(subset=['chla_ugl']).copy()
print(f"In-situ shallow data shape: {in_situ_shallow.shape}")
```

```python
# Ensure satellite date is datetime (it should be)
satellite['date'] = pd.to_datetime(satellite['date'])

# Merge
merged_shallow = pd.merge(
    in_situ_shallow[['lake_code', 'date', 'chla_ugl']],
    satellite[['lake_code', 'date', 'chla_sat']],
    on=['lake_code', 'date'],
    how='inner'
)

print(f"Matched days (depth ≤0.5): {len(merged_shallow)}")
print(merged_shallow.head())
```

```python
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

merged_shallow = merged_shallow.dropna(subset=['chla_ugl', 'chla_sat'])

if len(merged_shallow) > 0:
    r2 = r2_score(merged_shallow['chla_ugl'], merged_shallow['chla_sat'])
    rmse = np.sqrt(mean_squared_error(merged_shallow['chla_ugl'], merged_shallow['chla_sat']))
    bias = (merged_shallow['chla_sat'] - merged_shallow['chla_ugl']).mean()
    n = len(merged_shallow)
    
    print(f"Number of matched days (≤0.5m): {n}")
    print(f"R²: {r2:.3f}")
    print(f"RMSE: {rmse:.2f} µg/L")
    print(f"Bias: {bias:.2f} µg/L")
    
    # Per-lake breakdown
    for lake in merged_shallow['lake_code'].unique():
        subset = merged_shallow[merged_shallow['lake_code'] == lake]
        r2_lake = r2_score(subset['chla_ugl'], subset['chla_sat'])
        print(f"{lake}: n={len(subset)}, R²={r2_lake:.3f}")
else:
    print("No matched days after depth filtering.")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt

# Use daily_shallow (already filtered to depth <=0.5 and daily averaged)
sugg_daily = daily_shallow[daily_shallow['lake_code'] == 'SUGG'].copy()

# Keep only necessary columns (depth excluded)
features = ['temp', 'do', 'par', 'fdom']   # removed 'depth'
target = 'chla_ugl'

# Remove rows where target is missing
sugg_daily = sugg_daily.dropna(subset=[target])
print(f"SUGG daily rows with chla_ugl: {len(sugg_daily)}")

# Check feature coverage
print("Feature non‑null counts:")
print(sugg_daily[features].count())

# Impute missing feature values with median
imputer = SimpleImputer(strategy='median')
sugg_daily[features] = imputer.fit_transform(sugg_daily[features])

# Now we have a complete dataset
X = sugg_daily[features]
y = sugg_daily[target]
dates = sugg_daily['date']

# Identify the satellite dates for SUGG
sat_dates = satellite[satellite['lake_code'] == 'SUGG']['date'].unique()
print(f"Satellite dates for SUGG: {len(sat_dates)}")

# Split: train on all dates NOT in satellite dates, test on satellite dates
train_mask = ~dates.isin(sat_dates)
test_mask = dates.isin(sat_dates)

X_train, X_test = X[train_mask], X[test_mask]
y_train, y_test = y[train_mask], y[test_mask]
dates_train, dates_test = dates[train_mask], dates[test_mask]

print(f"Training size: {len(X_train)}, Test size (satellite days): {len(X_test)}")

# Train XGBoost
model = xgb.XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)

# Predict on test set
y_pred = model.predict(X_test)

# Evaluate predictions against in‑situ test values
r2_insitu = r2_score(y_test, y_pred)
rmse_insitu = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"XGBoost vs in‑situ (test days): R²={r2_insitu:.3f}, RMSE={rmse_insitu:.2f}")

# Merge predictions with satellite data
pred_df = pd.DataFrame({
    'lake_code': 'SUGG',
    'date': dates_test,
    'chla_pred': y_pred
})

sat_sugg = satellite[satellite['lake_code'] == 'SUGG'].copy()
merged_pred = pd.merge(pred_df, sat_sugg[['date', 'chla_sat']], on='date', how='inner')
print(f"Matched days for XGBoost vs satellite: {len(merged_pred)}")

if len(merged_pred) > 0:
    r2_sat = r2_score(merged_pred['chla_sat'], merged_pred['chla_pred'])
    rmse_sat = np.sqrt(mean_squared_error(merged_pred['chla_sat'], merged_pred['chla_pred']))
    bias_sat = (merged_pred['chla_pred'] - merged_pred['chla_sat']).mean()
    print(f"XGBoost vs Satellite: R²={r2_sat:.3f}, RMSE={rmse_sat:.2f} µg/L, Bias={bias_sat:.2f} µg/L")
    
    plt.figure(figsize=(6,6))
    plt.scatter(merged_pred['chla_sat'], merged_pred['chla_pred'], alpha=0.6, edgecolors='k')
    max_val = max(merged_pred['chla_sat'].max(), merged_pred['chla_pred'].max())
    plt.plot([0, max_val], [0, max_val], 'r--')
    plt.xlabel('Satellite Chla (µg/L)')
    plt.ylabel('XGBoost Predicted Chla (µg/L)')
    plt.title(f'XGBoost vs Satellite (SUGG): R²={r2_sat:.2f}, RMSE={rmse_sat:.2f}')
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    print("No matches – check date alignment.")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt
import shap

# ------------------------------------------------------------
# 1. Hyperparameter tuning using GridSearchCV
# ------------------------------------------------------------
print("Starting hyperparameter tuning...")
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.7, 0.8, 0.9]
}
grid = GridSearchCV(
    xgb.XGBRegressor(random_state=42, eval_metric='rmse'),
    param_grid,
    cv=3,
    scoring='r2',
    verbose=1,
    n_jobs=-1
)
grid.fit(X_train, y_train)
print(f"Best parameters: {grid.best_params_}")
best_model = grid.best_estimator_

# ------------------------------------------------------------
# 2. Evaluate tuned model on test set (in-situ)
# ------------------------------------------------------------
y_pred_tuned = best_model.predict(X_test)
r2_insitu_tuned = r2_score(y_test, y_pred_tuned)
rmse_insitu_tuned = np.sqrt(mean_squared_error(y_test, y_pred_tuned))
print(f"\nTuned XGBoost vs in-situ test days: R² = {r2_insitu_tuned:.3f}, RMSE = {rmse_insitu_tuned:.2f} µg/L")

# ------------------------------------------------------------
# 3. Merge tuned predictions with satellite data
# ------------------------------------------------------------
pred_df_tuned = pd.DataFrame({
    'lake_code': 'SUGG',
    'date': dates_test,
    'chla_pred_tuned': y_pred_tuned
})
sat_sugg = satellite[satellite['lake_code'] == 'SUGG'].copy()
merged_pred_tuned = pd.merge(pred_df_tuned, sat_sugg[['date', 'chla_sat']], on='date', how='inner')
print(f"Matched days for tuned model vs satellite: {len(merged_pred_tuned)}")

if len(merged_pred_tuned) > 0:
    r2_sat_tuned = r2_score(merged_pred_tuned['chla_sat'], merged_pred_tuned['chla_pred_tuned'])
    rmse_sat_tuned = np.sqrt(mean_squared_error(merged_pred_tuned['chla_sat'], merged_pred_tuned['chla_pred_tuned']))
    bias_sat_tuned = (merged_pred_tuned['chla_pred_tuned'] - merged_pred_tuned['chla_sat']).mean()
    print(f"Tuned XGBoost vs Satellite: R² = {r2_sat_tuned:.3f}, RMSE = {rmse_sat_tuned:.2f} µg/L, Bias = {bias_sat_tuned:.2f} µg/L")

    # Scatter plot
    plt.figure(figsize=(6,6))
    plt.scatter(merged_pred_tuned['chla_sat'], merged_pred_tuned['chla_pred_tuned'], alpha=0.6, edgecolors='k')
    max_val = max(merged_pred_tuned['chla_sat'].max(), merged_pred_tuned['chla_pred_tuned'].max())
    plt.plot([0, max_val], [0, max_val], 'r--')
    plt.xlabel('Satellite Chla (µg/L)')
    plt.ylabel('Tuned XGBoost Predicted Chla (µg/L)')
    plt.title(f'Tuned XGBoost vs Satellite (SUGG): R²={r2_sat_tuned:.2f}, RMSE={rmse_sat_tuned:.2f}')
    plt.grid(True, alpha=0.3)
    plt.show()
else:
    print("No matches – check date alignment.")

# ------------------------------------------------------------
# 4. SHAP analysis for interpretability
# ------------------------------------------------------------
print("\nComputing SHAP values...")
explainer = shap.TreeExplainer(best_model)
shap_values = explainer.shap_values(X_test)

# Summary plot (global feature importance)
shap.summary_plot(shap_values, X_test, feature_names=features, show=False)
plt.title('SHAP Feature Importance for XGBoost (test set)')
plt.tight_layout()
plt.savefig('shap_summary.png', dpi=150, bbox_inches='tight')
plt.show()

# Optional: Bar plot of mean absolute SHAP values
shap.summary_plot(shap_values, X_test, feature_names=features, plot_type='bar', show=False)
plt.title('Mean |SHAP| Feature Importance')
plt.tight_layout()
plt.savefig('shap_bar.png', dpi=150, bbox_inches='tight')
plt.show()

# ------------------------------------------------------------
# 5. Compare original vs tuned model performance (optional)
# ------------------------------------------------------------
print("\n--- Performance Comparison ---")
print(f"Original XGBoost vs in-situ test: R² = {r2_insitu:.3f}, RMSE = {rmse_insitu:.2f}")
print(f"Tuned XGBoost vs in-situ test:   R² = {r2_insitu_tuned:.3f}, RMSE = {rmse_insitu_tuned:.2f}")
```
